```python
import os
import subprocess

PROJECT_DIR = "/Users/atharvajayawant/Desktop/ache_docking"
VINA_BIN = os.path.join(PROJECT_DIR, "vina")

RECEPTOR = os.path.join(PROJECT_DIR, "4EY4.pdbqt")
LIGAND = os.path.join(PROJECT_DIR, "donepezil.pdbqt")
OUTPUT = os.path.join(PROJECT_DIR, "donepezil_docking.pdbqt")
```

```python
CENTER = (11.1, -53.7, 30.5)
SIZE = (20.0, 20.0, 20.0)

ENERGY_RANGE = 4
EXHAUSTIVENESS = 8
NUM_MODES = 20
```

```python
config_path = os.path.join(PROJECT_DIR, "vina_config.txt")

config = f"""receptor = {RECEPTOR}
ligand = {LIGAND}

center_x = {CENTER[0]}
center_y = {CENTER[1]}
center_z = {CENTER[2]}

size_x = {SIZE[0]}
size_y = {SIZE[1]}
size_z = {SIZE[2]}

energy_range = {ENERGY_RANGE}
exhaustiveness = {EXHAUSTIVENESS}
num_modes = {NUM_MODES}
"""

with open(config_path, "w") as f:
    f.write(config)

print(config)
```

```python
cmd = [
    VINA_BIN,
    "--config", config_path,
    "--out", OUTPUT
]

result = subprocess.run(cmd, capture_output=True, text=True)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
else:
    print(f"\nDocking completed successfully.")
    print(f"Output: {OUTPUT}")
    
```
